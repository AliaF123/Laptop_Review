```python
#Assignment- Section A
# 1. What is the number of unique values in the column "location_key"?
unique_values_df=covid_df['location_key'].nunique()
print(unique_values_df)
```

```python
# 2. What's the earliest date for which the data has been collected for? And the latest date? 
covid_df['date'] = pd.to_datetime(covid_df['date'])
earliest_date = covid_df['date'].min()
latest_date = covid_df['date'].max()
print("Earliest Date:", earliest_date)
print("Latest Date:", latest_date)
```

```python
# 3. On how many dates did we see the reported deaths greater than 100? 
deceased_100= covid_df[covid_df['new_deceased']>100]
ndates_df=deceased_100['date'].nunique()
print(ndates_df)

#len(covid_df[covid_df['new_deceased']>100])
```
